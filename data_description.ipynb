# Data Description – Netflix Dataset

This notebook describes the cleaned Netflix dataset.
The goal is to select a useful field for prediction, calculate
statistical values, create a graph and explain whether scaling is needed.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("netflix_titles_clean.csv", sep=";")
df.head()
```

## 2.1 Prediction Field

For this project, the field **release_year** was selected.
This field describes the year in which a movie or TV show was released.
It is suitable for analysis because it can show trends and changes over time.
It can also be compared with other fields such as type, rating and duration.

```python
df["release_year"].describe()
```

```python
print("Mean:", df["release_year"].mean())
print("Median:", df["release_year"].median())
print("Standard deviation:", df["release_year"].std())
print("Minimum:", df["release_year"].min())
print("Maximum:", df["release_year"].max())
```

## 2.2 Statistical Values

The selected field **release_year** was analysed using several statistical values.
The mean and median show the average release year of the titles.
The minimum and maximum values define the range of the data.
The standard deviation shows how strongly the release years are spread
around the average value.

```python
plt.figure(figsize=(10,6))

plt.hist(df["release_year"], bins=20, color="red", edgecolor="black")

plt.title("Distribution of Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid(True)

plt.show()
```

## 2.3 Graph

The histogram shows the distribution of Netflix titles by release year.
It helps to identify periods in which especially many titles were released.
This makes trends in the dataset visible and gives a better overview
of how the release years are distributed.

## 2.4 Scaling

Scaling is not necessary in this step.
Most of the fields in the dataset are categorical, such as type, country
or listed_in. The selected field **release_year** is already numeric and
has a meaningful scale. Therefore, no additional scaling method was
applied for this descriptive analysis.
